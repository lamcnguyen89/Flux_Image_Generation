## Run Pipeline

```python
# Edit these settings, then run this cell and the pipeline cell below.
from pathlib import Path

# Find the project from either its root or a notebook's nested working directory.
project_root = next(
    (parent for parent in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (parent / "Models" / "Model_Families").is_dir()),
    None,
)
if project_root is None:
    raise FileNotFoundError("Start the notebook kernel inside the Image_Generator project.")
model_family = "PonyDiffusionV6XL"
family_dir = project_root / "Models" / "Model_Families" / model_family
checkpoint_root = family_dir / "checkpoints"
lora_root = family_dir / "loras"

def resolve_path(value, relative_to):
    """Resolve relative paths against an explicit root; preserve absolute paths."""
    path = Path(value).expanduser()
    return (path if path.is_absolute() else relative_to / path).resolve()

prompts = [
    "A cat holding a sign that says hello world",
    "A cinematic photograph of a mountain lake at sunrise, mist over the water, golden light",
    "A friendly robot tending a rooftop greenhouse, watercolor illustration, soft pastel colors",
    "A glowing bioluminescent forest at night, tiny fireflies, intricate fantasy concept art",
]

# Official SDXL 1.0 base repository also supplies config/tokenizers for local checkpoints.
model_id = "stabilityai/stable-diffusion-xl-base-1.0"
# Path relative to this family's checkpoints/, or an absolute path. None = official base model.
# Rerun this settings cell and the pipeline cell after changing models or LoRAs.
checkpoint_file = "ponyDiffusionV6XL_v6StartWithThisOne.safetensors"
height = 1024                 # Pixels; use multiples of 8. SDXL is trained for 1024x1024.
width = 1024
num_inference_steps = 50       # Try 20 for faster previews.
guidance_scale = 5.0           # SDXL classifier-free guidance; > 1 enables negative prompting.
num_images_per_prompt = 1      # Images generated together; larger batches use more memory.
seed = None                      # None = random seeds; an integer = repeatable seed sequence.

# Negative prompts are used when guidance_scale > 1.
# SDXL's CLIP text encoders have a 77-token context; longer prompts are truncated.
negative_prompt = ""

# No local LoRA is available. Select one from this family's loras/ before enabling it.
Use_LoRA = False
LoRA_Weights = None  # Relative to this family's loras/, or an absolute path.
LoRA_Scale = 0.8                     # 0 = no influence, 1 = full trained strength.
# Add the selected LoRA's trigger words to your prompts if its author requires them.

# Relative input/output paths are anchored at the project root.
output_dir = "Images"
show_images = True            # Display generated images inline as well as saving them.

```

```python
import gc
import hashlib
import json
import math
import os
from pathlib import Path
import re
import secrets
from time import perf_counter
from uuid import uuid4

from dotenv import load_dotenv
from IPython.display import display
import torch
from diffusers import StableDiffusionXLPipeline

# Validate settings before downloading or loading model weights.
if not isinstance(prompts, (list, tuple)) or not prompts:
    raise ValueError("prompts must be a nonempty list of strings.")
if any(not isinstance(prompt, str) or not prompt.strip() for prompt in prompts):
    raise ValueError("Every prompt must be a nonempty string.")
unique_prompts = list(dict.fromkeys(prompts))  # Preserve order and skip exact duplicates.
for name, value in {"height": height, "width": width}.items():
    if type(value) is not int or value <= 0 or value % 8:
        raise ValueError(f"{name} must be a positive integer divisible by 8.")
for name, value in {
    "num_inference_steps": num_inference_steps,
    "num_images_per_prompt": num_images_per_prompt,
}.items():
    if type(value) is not int or value <= 0:
        raise ValueError(f"{name} must be a positive integer.")
if seed is not None and (type(seed) is not int or not 0 <= seed < 2**63):
    raise ValueError("seed must be None or an integer between 0 and 2**63 - 1.")
if isinstance(guidance_scale, bool) or not isinstance(guidance_scale, (int, float)) or not math.isfinite(guidance_scale) or guidance_scale < 0:
    raise ValueError("guidance_scale must be a finite number >= 0.")
if not isinstance(negative_prompt, str):
    raise ValueError("negative_prompt must be a string (use an empty string to leave it blank).")


# Validate and fingerprint the optional checkpoint before loading or releasing any model.
checkpoint_path = None
checkpoint_key = None
if checkpoint_file is not None:
    if not isinstance(checkpoint_file, (str, Path)) or not str(checkpoint_file).strip():
        raise ValueError("Set checkpoint_file to a local SDXL .safetensors file, or None for the base model.")
    checkpoint_path = resolve_path(checkpoint_file, checkpoint_root)
    if not checkpoint_path.is_file():
        raise FileNotFoundError(f"Checkpoint not found: {checkpoint_path}")
    if checkpoint_path.suffix.lower() != ".safetensors":
        raise ValueError("Select a local .safetensors checkpoint.")
    checkpoint_stat = checkpoint_path.stat()
    checkpoint_key = (
        str(checkpoint_path), checkpoint_stat.st_size, checkpoint_stat.st_mtime_ns
    )

# Check local LoRA settings before loading weights or changing the active adapter.
if type(Use_LoRA) is not bool:
    raise ValueError("Use_LoRA must be True or False.")
if isinstance(LoRA_Scale, bool) or not isinstance(LoRA_Scale, (int, float)) or not math.isfinite(LoRA_Scale) or LoRA_Scale < 0:
    raise ValueError("LoRA_Scale must be a finite number >= 0.")
lora_path = None
lora_key = None
if Use_LoRA:
    if not isinstance(LoRA_Weights, (str, Path)) or not str(LoRA_Weights).strip():
        raise ValueError("Set LoRA_Weights to a local SDXL LoRA .safetensors file.")
    lora_path = resolve_path(LoRA_Weights, lora_root)
    if not lora_path.is_file():
        raise FileNotFoundError(f"LoRA file not found: {lora_path}")
    if lora_path.suffix.lower() != ".safetensors":
        raise ValueError("Select a local .safetensors LoRA file.")
    lora_stat = lora_path.stat()
    lora_key = (str(lora_path), lora_stat.st_size, lora_stat.st_mtime_ns)

load_dotenv(project_root / ".env")
if not torch.cuda.is_available():
    raise RuntimeError("CUDA is unavailable. Select the huggingface kernel with the Thor CUDA build of PyTorch.")

device = torch.device("cuda:0")
# SDXL automatically upcasts its VAE to float32 when required for stable decoding.
dtype = torch.float16
gib = 1024**3
free_bytes, total_bytes = torch.cuda.mem_get_info(device)
print(f"GPU: {torch.cuda.get_device_name(device)} | CUDA: {torch.version.cuda}")
print(f"CUDA memory: {free_bytes / gib:.1f} GiB free / {total_bytes / gib:.1f} GiB total")

# CACHE_LOCATION is the HF_HOME root; retain its hub/ layout for existing downloads.
cache_root = resolve_path(os.getenv("CACHE_LOCATION") or "Models", project_root)
cache_dir = cache_root / "hub"
cache_dir.mkdir(parents=True, exist_ok=True)

# Reuse the model when only prompts or generation settings change.
# Thor's CUDA allocations use its shared RAM; keep all components on the GPU.
# Reload when the selected file changes, including replacement at the same path.
pipeline_config = (
    model_id, checkpoint_key, str(cache_dir.resolve()), dtype, str(device)
)
if "pipe" not in globals() or globals().get("_loaded_sdxl_config") != pipeline_config:
    if "pipe" in globals():
        del pipe
    gc.collect()
    torch.cuda.empty_cache()
    if checkpoint_path is None:
        pipe = StableDiffusionXLPipeline.from_pretrained(
            model_id,
            cache_dir=str(cache_dir),
            token=os.getenv("HUGGINGFACE_TOKEN") or None,
            dtype=dtype,
            variant="fp16",
            use_safetensors=True,
            low_cpu_mem_usage=True,
        ).to(device)
    else:
        # SDXL single-file checkpoints include the UNet, text encoders, and VAE.
        pipe = StableDiffusionXLPipeline.from_single_file(
            str(checkpoint_path),
            config=model_id,
            cache_dir=str(cache_dir),
            token=os.getenv("HUGGINGFACE_TOKEN") or None,
            dtype=dtype,
            use_safetensors=True,
        ).to(device)
    _loaded_sdxl_config = pipeline_config
else:
    print("Reusing the loaded pipeline.")
print(f"Checkpoint: {checkpoint_path.name}" if checkpoint_path else f"Base model: {model_id}")


# Keep the adapter separate from base weights so switching/off restores the base model.
# The file fingerprint also reloads weights replaced at the same path.
if getattr(pipe, "_notebook_lora_key", None) != lora_key:
    pipe.unload_lora_weights()
    pipe._notebook_lora_key = None
    if Use_LoRA:
        try:
            pipe.load_lora_weights(
                str(lora_path.parent),
                weight_name=lora_path.name,
                adapter_name="notebook_lora",
                local_files_only=True,
            )
        except Exception as exc:
            # A failed load must not leave a partially applied adapter for the next run.
            pipe.unload_lora_weights()
            raise RuntimeError(
                f"Could not load LoRA {lora_path.name!r}. Check that it matches {model_id}."
            ) from exc
        pipe._notebook_lora_key = lora_key

if Use_LoRA:
    pipe.set_adapters("notebook_lora", adapter_weights=float(LoRA_Scale))
    pipe.set_lora_device(adapter_names=["notebook_lora"], device=device)
    print(f"LoRA: {lora_path.name} | Strength: {LoRA_Scale}")
else:
    print("LoRA disabled; using the selected model without an adapter.")
lora_settings = {
    "enabled": Use_LoRA,
    "weights": str(lora_path) if Use_LoRA else None,
    "scale": float(LoRA_Scale) if Use_LoRA else 0.0,
}

for name in ("unet", "text_encoder", "text_encoder_2", "vae"):
    component = getattr(pipe, name)
    print(f"{name}: {component.device}, {component.dtype}")
del component

output_root = resolve_path(output_dir, project_root)
output_root.mkdir(parents=True, exist_ok=True)
saved_paths = []
torch.cuda.reset_peak_memory_stats(device)

# Process prompts in sequence; num_images_per_prompt controls the batch size.
with torch.inference_mode():
    for prompt_index, prompt in enumerate(unique_prompts):
        # The hash distinguishes different prompts with the same sanitized name.
        slug = re.sub(r"[^a-z0-9]+", "-", prompt.lower()).strip("-")[:60].rstrip("-") or "prompt"
        prompt_hash = hashlib.sha256(prompt.encode("utf-8")).hexdigest()[:16]
        prompt_dir = output_root / f"{slug}-{prompt_hash}"
        prompt_dir.mkdir(parents=True, exist_ok=True)
        (prompt_dir / "prompt.txt").write_text(prompt, encoding="utf-8")

        # Give each image its own recorded seed, including when seeds are random.
        seeds = [
            secrets.randbelow(2**63) if seed is None
            else (seed + prompt_index * num_images_per_prompt + image_index) % (2**63)
            for image_index in range(num_images_per_prompt)
        ]
        generators = [torch.Generator(device=device).manual_seed(value) for value in seeds]
        generation_settings = {
            "height": height,
            "width": width,
            "num_inference_steps": num_inference_steps,
            "guidance_scale": guidance_scale,
            "num_images_per_prompt": num_images_per_prompt,
            "negative_prompt": negative_prompt,
        }
        print(f"Prompt {prompt_index + 1}/{len(unique_prompts)}: {prompt}")
        torch.cuda.synchronize(device)
        start = perf_counter()
        images = pipe(prompt=prompt, generator=generators, **generation_settings).images
        torch.cuda.synchronize(device)
        elapsed = perf_counter() - start

        for image, image_seed in zip(images, seeds, strict=True):
            # A fresh UUID preserves prior images, even when rerunning the same seed.
            output_path = prompt_dir / f"sdxl-seed-{image_seed}-{uuid4().hex}.png"
            image.save(output_path)
            metadata = {
                "prompt": prompt,
                "model_id": model_id,
                "checkpoint": str(checkpoint_path) if checkpoint_path else None,
                "lora": dict(lora_settings),
                "seed": image_seed,
                "dtype": str(dtype),
                "device": str(device),
                "torch_version": torch.__version__,
                "batch_generation_seconds": elapsed,
                **generation_settings,
            }
            output_path.with_suffix(".json").write_text(
                json.dumps(metadata, indent=2, ensure_ascii=False), encoding="utf-8"
            )
            saved_paths.append(output_path)
            print(f"Saved {output_path}")
            if show_images:
                display(image)
        print(f"Generation: {elapsed:.1f}s for {len(images)} image(s)")

print(f"Saved {len(saved_paths)} images across {len(unique_prompts)} prompt folders.")
print(f"Peak PyTorch CUDA allocation: {torch.cuda.max_memory_allocated(device) / gib:.1f} GiB")
```
